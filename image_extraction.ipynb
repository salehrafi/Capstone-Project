```python
import os
import shutil
from pathlib import Path
```

```python
SOURCE_FOLDER = r"E:\Capstone Dataset\Chest -20260226T162017Z-1-001\Chest\Cronic Lung Disease\Emphysematous change"   # source folder
OUTPUT_FOLDER = r"E:\Capstone Dataset\Chest -20260226T162017Z-1-001\Chest\Extracted image cronic lung desise"  # Destination Folder
```

```python
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".gif", ".bmp", ".webp", ".svg", ".tiff", ".tif"}

# Duplicate file name occure hoile ki korbe: 'rename' or 'skip'
ON_DUPLICATE = "rename"


def extract_images():
    source = Path(SOURCE_FOLDER)
    output = Path(OUTPUT_FOLDER)

    # Validation
    if not source.exists():
        print(f"Source folder not exixsts: {source}")
        return

    # Output folder if not exists 
    output.mkdir(parents=True, exist_ok=True)
    print(f" Source:  {source}")
    print(f" Output:  {output}")
    print("-" * 50)

    copied = 0
    skipped = 0
    renamed = 0

    # go through all subfolder
    for file_path in source.rglob("*"):
        if file_path.is_file() and file_path.suffix.lower() in IMAGE_EXTENSIONS:

            dest = output / file_path.name

            # Duplicate handling
            if dest.exists():
                if ON_DUPLICATE == "skip":
                    print(f"⏭  Skip (already exists): {file_path.name}")
                    skipped += 1
                    continue
                elif ON_DUPLICATE == "rename":
                    # new name:  image(1).jpg, image(2).jpg ...
                    stem = file_path.stem
                    suffix = file_path.suffix
                    counter = 1
                    while dest.exists():
                        dest = output / f"{stem}({counter}){suffix}"
                        counter += 1
                    print(f" Rename → {dest.name}  (from: {file_path.parent.name})")
                    renamed += 1

            shutil.copy2(file_path, dest)
            if dest.name == file_path.name:
                print(f"Copied: {file_path.name}  <-  .../{file_path.parent.name}/")
            copied += 1
            
            # Summary
            print("-" * 50)
            print(f"Done!")
            print(f" Copied  : {copied} images")
            print(f"Renamed : {renamed} duplicate (dplicate was)")
            print(f"Skipped : {skipped}")
            print(f"Output  : {output}")
```

```python
if __name__ == "__main__":
    extract_images()
```
